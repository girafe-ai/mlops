## Launching a distributed job

Throughout the seminar we launch scripts with `torchrun`. The helper in [launch.sh](launch.sh) wraps the usual invocation:

```bash
./launch.sh 2 ../01_baseline/train.py        # single process, ignored NGPUS for the baseline
./launch.sh 2 ../02_data_parallel/train_ddp.py
```

`torchrun` sets `RANK`, `WORLD_SIZE`, and `LOCAL_RANK` in each worker's environment. Every script in the seminar consumes them through [dist_utils.py](dist_utils.py).
